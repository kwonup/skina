# skina 데이터 EDA

AI-Hub 원본 데이터의 클래스별 이미지 수, 이미지 크기와 샘플을 확인합니다. 이 노트북은 데이터를 수정하지 않습니다.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from PIL import Image

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RAW_DIR = PROJECT_ROOT / 'data' / 'raw'
IMAGE_EXTENSIONS = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff', '.webp'}
PROJECT_ROOT
```

## 클래스별 이미지 수

```python
rows = []
for split in ['train', 'validation']:
    split_dir = RAW_DIR / split
    for class_dir in sorted(path for path in split_dir.iterdir() if path.is_dir()):
        count = sum(
            path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
            for path in class_dir.iterdir()
        )
        rows.append({'split': split, 'class_name': class_dir.name, 'count': count})

counts = pd.DataFrame(rows)
counts.pivot(index='class_name', columns='split', values='count')
```

```python
counts.groupby('split')['count'].sum()
```

## 클래스별 샘플 이미지

```python
class_dirs = sorted(path for path in (RAW_DIR / 'train').iterdir() if path.is_dir())
fig, axes = plt.subplots(3, 5, figsize=(16, 10))

for axis, class_dir in zip(axes.flat, class_dirs):
    image_path = next(
        path for path in sorted(class_dir.iterdir())
        if path.is_file() and path.suffix.lower() in IMAGE_EXTENSIONS
    )
    with Image.open(image_path) as image:
        axis.imshow(image.convert('RGB'))
        axis.set_title(f'{class_dir.name}\n{image.size}')
    axis.axis('off')

plt.tight_layout()
plt.show()
```
